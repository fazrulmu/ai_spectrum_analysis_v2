# Visualisasi Spektrum dari Dataset Gabungan

Notebook ini bertujuan untuk memvisualisasikan data spektrum yang sudah diproses dari file `dataset.csv`. Alur kerja ini lebih efisien karena kita tidak perlu memproses ulang file mentah setiap kali ingin melihat data.

## 1. Pengaturan Awal

Pertama, kita akan mengimpor semua pustaka yang diperlukan dan mengatur path proyek.

```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Menambahkan root proyek ke sys.path agar bisa diakses dari mana saja
project_root_path = Path.cwd().parent
if str(project_root_path) not in sys.path:
    sys.path.insert(0, str(project_root_path))

print(f"✅ Root proyek diatur ke: {project_root_path}")
```

## 2. Memuat Dataset Training (Format Lebar)

Kita akan memuat file `training_dataset.csv` yang berisi semua data spektrum dalam format 'lebar' yang siap untuk training.

```python
DATASET_PATH = project_root_path / 'data' / 'for_train' / 'training_dataset.csv'
#data/standarize/processed_spectrum
if DATASET_PATH.exists():
    df_dataset = pd.read_csv(DATASET_PATH)
    print(f"✅ Dataset training berhasil dimuat. Total baris: {len(df_dataset)}")
    print(f"Jumlah molekul unik: {df_dataset['molecule_id'].nunique()}")
    display(df_dataset.head())
else:
    print(f"❌ File dataset tidak ditemukan di {DATASET_PATH}. Pastikan skrip 'create_training_dataset.py' sudah dijalankan.")
```

## 3. Visualisasi Spektrum IR Spesifik

Sekarang kita akan memilih satu molekul dari dataset training, mengubahnya kembali ke format 'panjang', dan memplotnya.

```python
# Tentukan ID molekul yang ingin divisualisasikan dari training_dataset.csv
target_molecule_id = 'IR_529-34-0_IR_gas_0.csv' # Ganti dengan ID lain jika perlu

# Cek apakah DataFrame ada dan tidak kosong
if 'df_dataset' in locals() and not df_dataset.empty and target_molecule_id in df_dataset['molecule_id'].values:
    # 1. Ambil satu baris data untuk molekul target
    spectrum_series = df_dataset[df_dataset['molecule_id'] == target_molecule_id].iloc[0]
    
    # 2. Pisahkan kolom fitur (bin) dari kolom lain
    feature_cols = [col for col in df_dataset.columns if col.startswith('bin_')]
    spectrum_data = spectrum_series[feature_cols]
    
    # 3. Ubah kembali ke format panjang: ekstrak wavenumber (x) dan absorbance (y)
    # Ambil angka dari nama kolom 'bin_xxxx' sebagai nilai x
    x_values = [int(col.split('_')[1]) for col in spectrum_data.index]
    y_values = spectrum_data.values
    
    # 4. Plotting
    plt.figure(figsize=(15, 7))
    plt.plot(x_values, y_values, label=f"Spektrum IR untuk {target_molecule_id}")
    plt.title(f"Spektrum IR - {target_molecule_id}", fontsize=16)
    plt.xlabel("Wavenumber (cm⁻¹)")
    plt.ylabel("Absorbance (Normalized)")
    plt.gca().invert_xaxis()  # Balik sumbu x untuk konvensi IR
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.tight_layout()
    plt.show()
else:
    print(f"❌ Tidak ditemukan data untuk molecule_id '{target_molecule_id}' atau DataFrame 'df_dataset' tidak ada.")
```

## 4. Kesimpulan

Notebook ini menunjukkan alur kerja yang benar dan efisien untuk memvisualisasikan data dari dataset gabungan. Dengan memuat satu file CSV dan memfilternya, kita dapat dengan cepat memplot spektrum apa pun tanpa perlu memproses ulang data mentah.

## 5. Visualisasi Spektrum dari File CSV Individual (Metode Alternatif)

Metode ini lebih sederhana jika Anda hanya ingin memvisualisasikan satu file spektrum yang sudah diproses (misalnya, dari direktori `data/standarize/processed_spectrum/`). Kita akan memuat file CSV tersebut secara langsung.

```python
# 1. Tentukan path ke file CSV yang spesifik
single_csv_path = project_root_path / 'data' / 'standarize' / 'processed_spectrum' / 'IR_0_563-54-2.csv'

# 2. Muat file CSV ke dalam DataFrame data/standarize/processed_spectrum/IR_10024-97-2_IR_GAS_(100_mmHg,_N2_ADDED,_TOTAL_PRESSURE_600_mmHg)_1.csv
if single_csv_path.exists():
    df_single = pd.read_csv(single_csv_path)
    molecule_id = df_single['molecule_id'].iloc[0]
    
    # 3. Plotting (lebih sederhana karena data sudah dalam format panjang)
    plt.figure(figsize=(15, 7))
    plt.plot(df_single['x_value'], df_single['y_value'], label=f"Spektrum IR untuk {molecule_id}", color='purple')
    plt.title(f"Spektrum IR dari File Individual - {molecule_id}", fontsize=16)
    plt.xlabel("Wavenumber (cm⁻¹)")
    plt.ylabel("Absorbance (Normalized)")
    plt.gca().invert_xaxis()  # Balik sumbu x untuk konvensi IR
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.tight_layout()
    plt.show()
else:
    print(f"❌ File spektrum individual tidak ditemukan di {single_csv_path}")
```

Metode kedua ini sangat berguna untuk memeriksa hasil dari skrip `preprocess_universal.py` secara cepat.
